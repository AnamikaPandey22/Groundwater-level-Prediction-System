```python
import os
from google.colab import drive
import pandas as pd

drive.mount('/content/drive')

# Folder path in Google Drive (update if folder name is different)
folder_path = "/content/drive/MyDrive/gandak_groundwater"

# List all files to verify
print("Files in folder:", os.listdir(folder_path))

# Initialize list to store dataframes
df_list = []

# Loop through CSV files and merge
for file in os.listdir(folder_path):
    if file.endswith(".csv"):
        file_path = os.path.join(folder_path, file)
        district_name = file.replace(".csv", "")  # Extract district name from filename
        df = pd.read_csv(file_path)
        df['district_name'] = district_name  # Add district column
        df_list.append(df)

# Concatenate all dataframes
merged_df = pd.concat(df_list, ignore_index=True)

# Save the merged master CSV
merged_df.to_csv("/content/gandak_groundwater_master.csv", index=False)

print("✅ All CSVs merged into 'gandak_groundwater_master.csv'")
print("Rows:", merged_df.shape[0], "Columns:", merged_df.shape[1])

# Show first 5 rows
merged_df.head()
```

```python
import pandas as pd
from prophet import Prophet
import matplotlib.pyplot as plt
import joblib

# Load dataset
df = pd.read_csv("/content/gandak_groundwater_master.csv", low_memory=False)

# Filter for one district (example: Bettiah2)
district = 'Bettiah2'
df_district = df[df['district_name'] == district]

# Prepare data for Prophet
data = df_district[['DateTime', 'mbgl']].copy()  # Using 'mbgl' as groundwater level
data.columns = ['ds', 'y']

# Convert DateTime to proper datetime format
data['ds'] = pd.to_datetime(data['ds'], dayfirst=True)  # Because your format is dd/mm/yyyy

# Initialize and fit Prophet model
model = Prophet()
model.fit(data)

# Predict next 30 days
future = model.make_future_dataframe(periods=30, freq='D')
forecast = model.predict(future)

# Show forecast
print(forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail())

# Plot forecast
fig = model.plot(forecast)
plt.title(f"Groundwater Forecast for {district}")
plt.show()

# Save model for chatbot
joblib.dump(model, f"gw_model_{district}.pkl")
print(f"✅ Prophet model saved as gw_model_{district}.pkl")
```

```python
import pandas as pd
from prophet import Prophet
import joblib

# Load dataset
df = pd.read_csv("/content/gandak_groundwater_master.csv", low_memory=False)

# Unique districts
districts = df['district_name'].unique()
print("Districts:", districts)

for district in districts:
    df_district = df[df['district_name'] == district]

    # Select date + groundwater level
    data = df_district[['DateTime', 'mbgl']].copy()
    data.columns = ['ds', 'y']
    data['ds'] = pd.to_datetime(data['ds'], dayfirst=True)

    # Drop rows with NaN groundwater
    data = data.dropna(subset=['y'])

    if data.shape[0] < 2:
        print(f"⚠️ Skipping {district}: not enough data ({data.shape[0]} rows)")
        continue

    # Train Prophet
    model = Prophet()
    model.fit(data)

    # Save model
    joblib.dump(model, f"gw_model_{district}.pkl")
    print(f"✅ Model saved for {district}")

```

```python
import joblib
import os

# Folder where all models are saved
model_folder = "/content/"  # update if your models are in a different folder

# Find all .pkl files
model_files = [f for f in os.listdir(model_folder) if f.startswith("gw_model_") and f.endswith(".pkl")]

# Preload models into a dictionary
district_models = {}
for file in model_files:
    district_name = file.replace("gw_model_", "").replace(".pkl", "")
    district_models[district_name] = joblib.load(os.path.join(model_folder, file))
    print(f"✅ Loaded model for {district_name}")

print("\nAll district models preloaded successfully!")
```

```python
district = "Bagaha2"

if district in district_models:
    model = district_models[district]

    # Predict next 30 days
    future = model.make_future_dataframe(periods=30, freq='D')
    forecast = model.predict(future)

    latest = forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail(1)
    print(f"Groundwater prediction for {district}:")
    print(latest)
else:
    print(f"⚠️ No model available for {district}")
```
