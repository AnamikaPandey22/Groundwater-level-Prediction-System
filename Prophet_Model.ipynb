```python
import pandas as pd
from prophet import Prophet
import os

# --- Step 1: Load Dataset ---
file_path = '/content/drive/MyDrive/Github groundwater dataset/groundwater_Report of Trends of Water Level_Bihar.xls'
df = pd.read_excel(file_path, skiprows=4)  # skip first 4 header rows

# --- Step 2: Clean & Prepare Data ---
# Select only relevant columns
df = df[['State', 'District', 'Pre-Monsoon', 'Post-Monsoon']].copy()

# Drop rows with missing District
df = df.dropna(subset=['District'])

# Convert Pre-Monsoon and Post-Monsoon to numeric
df['Pre-Monsoon'] = pd.to_numeric(df['Pre-Monsoon'], errors='coerce')
df['Post-Monsoon'] = pd.to_numeric(df['Post-Monsoon'], errors='coerce')

# --- Step 3: Create folder to save models ---
models_folder = '/content/groundwater_models'
os.makedirs(models_folder, exist_ok=True)

# --- Step 4: Train Prophet models for each District ---
districts = df['District'].unique()
for district in districts:
    print(f"Training model for {district}...")

    # Filter data for district
    site_df = df[df['District'] == district][['Pre-Monsoon', 'Post-Monsoon']].dropna()

    # Prophet requires 'ds' (datetime) and 'y' (value) columns
    # We'll create yearly date range from 1994
    site_df = site_df.reset_index(drop=True)
    n_rows = len(site_df)
    if n_rows < 2:
        print(f"⚠️ Skipping {district}: not enough data ({n_rows} rows)")
        continue

    # Train separate models for Pre-Monsoon and Post-Monsoon
    for season in ['Pre-Monsoon', 'Post-Monsoon']:
        season_df = pd.DataFrame()
        season_df['ds'] = pd.date_range(start='1994', periods=n_rows, freq='Y')
        season_df['y'] = site_df[season].values

        model = Prophet(yearly_seasonality=True)
        model.fit(season_df)

        # Save the model
        model_name = f"{district}_{season.replace('-', '')}_model"
        model_path = os.path.join(models_folder, model_name)
        model.save(model_path)
        print(f"✅ Model saved for {district} ({season})")

print("🎉 All models trained and saved successfully!")
```

```python
# -----------------------------
# Step 1: Imports & Setup
# -----------------------------
import pandas as pd
import os
from prophet import Prophet
import joblib # Import joblib for saving the model

# -----------------------------
# Step 2: Load & Clean Dataset
# -----------------------------
# Use the uploaded file path
file_path = "/content/groundwater_Report of Trends of Water Level_Bihar.xls"

# Skip first 4 rows to get the correct header
df = pd.read_excel(file_path, header=4)

# Clean column names (strip spaces) - not needed with header=4
# df.columns = df.columns.str.strip()

# Select relevant columns based on the output of the previous successful run
df = df[['State', 'District', 'Pre-Monsoon', 'Post-Monsoon']]

# Drop rows with missing District
df = df.dropna(subset=['District'])

# Convert Pre/Post-Monsoon to numeric
df['Pre-Monsoon'] = pd.to_numeric(df['Pre-Monsoon'], errors='coerce')
df['Post-Monsoon'] = pd.to_numeric(df['Post-Monsoon'], errors='coerce')

# -----------------------------
# Step 3: Prepare Models Folder
# -----------------------------
models_folder = "/content/groundwater_models"
os.makedirs(models_folder, exist_ok=True)

# -----------------------------
# Step 4: Train Prophet Models
# -----------------------------
districts = df['District'].unique()
seasons = ['Pre-Monsoon', 'Post-Monsoon']

for district in districts:
    print(f"Training models for {district}...")
    district_df = df[df['District'] == district]

    for season in seasons:
        season_df = district_df[['State', 'District', season]].copy()
        season_df = season_df.rename(columns={season: 'y'})
        season_df = season_df.dropna(subset=['y'])

        n_rows = len(season_df)
        if n_rows < 2:
            print(f"⚠️ Skipping {district} ({season}): not enough data ({n_rows} rows)")
            continue

        # Create ds column (yearly) using 'YE' frequency
        season_df['ds'] = pd.date_range(start='1994', periods=n_rows, freq='YE')

        # Initialize and fit Prophet model
        model = Prophet(yearly_seasonality=True, weekly_seasonality=False, daily_seasonality=False)
        model.fit(season_df[['ds', 'y']])

        # Save model using joblib
        model_name = f"{district}_{season.replace('-', '')}_model.pkl"
        model_path = os.path.join(models_folder, model_name)
        joblib.dump(model, model_path)
        print(f"✅ Model saved for {district} ({season})")

print("✅ All models trained and saved successfully!")

# -----------------------------
# Step 5: Show Prediction Example
# -----------------------------
# Here is an example of how you would load a model and make a prediction:

# Choose a district and season to demonstrate
example_district = 'ARWAL' # Changed to a district with a trained model
example_season = 'Pre-Monsoon'

if example_district:
    print(f"\nShowing prediction example for {example_district} ({example_season})...")
    # Load the saved model
    example_model_name = f"{example_district}_{example_season.replace('-', '')}_model.pkl"
    example_model_path = os.path.join(models_folder, example_model_name)

    if os.path.exists(example_model_path):
        loaded_model = joblib.load(example_model_path)

        # Create a future dataframe for prediction
        future = loaded_model.make_future_dataframe(periods=5, freq='YE')

        # Make predictions
        forecast = loaded_model.predict(future)

        # Display the forecast
        print("\nExample Forecast:")
        display(forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail())
    else:
        print(f"❌ Model file not found for {example_district} ({example_season}).")
```

```python
from google.colab import files
import shutil
import os

# Upload the zipped folder containing the .pkl models
uploaded = files.upload()

# Assuming the uploaded file is a zip archive
for fn in uploaded.keys():
    zip_file_path = fn
    print(f'User uploaded file "{fn}" with length {len(uploaded[fn])} bytes')

# Define the destination folder
models_folder = "/content/groundwater_models"

# Create the destination folder if it doesn't exist
os.makedirs(models_folder, exist_ok=True)

# Unzip the uploaded file into the destination folder
try:
    shutil.unpack_archive(zip_file_path, models_folder)
    print(f"Successfully extracted models to {models_folder}")
except Exception as e:
    print(f"Error extracting file: {e}")

# Clean up the uploaded zip file
os.remove(zip_file_path)
```
